## Step 1 · The sounds

Each sound is one line with:
- `name` and `place`: what you see in the popup,
- `kind`: `music`, `street` or `monument` (it sets the colour of the marker),
- `lat` and `lon`: where it was recorded,
- `freesound`: the number of the sound on [freesound.org](https://freesound.org), a free library of sounds.

All six recordings are **CC0** (free for everyone) and were made by travellers in Uzbekistan.

## Step 2 · The map

For every sound we put a **marker** on the map. Its **popup** (the box that opens when you click) holds a small player.

- Freesound sounds use Freesound's own player (`<iframe>`).
- Your own recordings use the browser's player (`<audio>`).

Click a marker, then press ▶. Turn your sound on! 🔊

```python
COLORS = {"music": "orange", "street": "red", "monument": "cadetblue"}
ICONS = {"music": "music", "street": "people-group", "monument": "landmark"}

def player(s):
    if "audio" in s:   # your own recording
        return f'<audio controls style="width:300px" src="{s["audio"]}"></audio>'
    url = f'https://freesound.org/embed/sound/iframe/{s["freesound"]}/simple/medium/'
    return f'<iframe src="{url}" width="481" height="86" frameborder="0" scrolling="no"></iframe>'

def make_map():
    m = folium.Map(location=[40.6, 64.5], zoom_start=6, tiles="Esri.WorldGrayCanvas")
    for s in sounds:
        html = (f'<b>{s["name"]}</b><br>'
                f'<span style="color:#666">{s["place"]}</span><br>'
                f'{player(s)}<br>'
                f'<small>Sound: {s["author"]}</small>')
        folium.Marker(
            [s["lat"], s["lon"]],
            popup=folium.Popup(html, max_width=500),
            tooltip=s["name"],
            icon=folium.Icon(color=COLORS[s["kind"]], icon=ICONS[s["kind"]], prefix="fa"),
        ).add_to(m)
    return m

m = make_map()
m
```

## Step 3 · Add a sound from Freesound

1. Go to [freesound.org](https://freesound.org) and search, for example, **Khiva**, **Bukhara** or **Tashkent**.
2. Open a sound. Its address looks like `freesound.org/people/selcukartut/sounds/686051/`. The **number at the end** is what you need.
3. Check the **licence** on the right. CC0 and CC BY are the easiest to use.
4. Change the line below: the name, the place, the coordinates and the number. Run the cell, then run the map cell (Step 2) again.

> **Coordinates:** in [Google Maps](https://maps.google.com) or [OpenStreetMap](https://www.openstreetmap.org), right-click the place. The first number is `lat`, the second is `lon`.

```python
sounds.append(
    {"name": "Musicians in a restaurant", "place": "Khiva", "kind": "music",
     "lat": 41.3790, "lon": 60.3610, "freesound": 686051, "author": "selcukartut"}
)

m = make_map()
m
```

## Step 4 · Add your own recording 🎙️

1. Record 20–60 seconds with your phone: a bazaar, the azan, rain in your mahalla, a wedding, the metro. Send the file to your computer (**mp3** or **m4a**).
2. Run the cell below and click **Choose Files**. Pick your recording.
3. Write the name, the place and the coordinates of your recording in the next cell and run it.

The sound is stored **inside** the map file, so it works even without Freesound.

```python
import base64, os
from google.colab import files

uploaded = files.upload()
filename = list(uploaded)[0]

TYPES = {".mp3": "audio/mpeg", ".m4a": "audio/mp4", ".wav": "audio/wav",
         ".ogg": "audio/ogg", ".aac": "audio/aac"}
mime = TYPES.get(os.path.splitext(filename)[1].lower(), "audio/mpeg")
my_audio = f"data:{mime};base64," + base64.b64encode(uploaded[filename]).decode()

print("Loaded", filename, round(len(uploaded[filename]) / 1e6, 1), "MB ✅")
```

```python
sounds.append(
    {"name": "My recording", "place": "Tashkent", "kind": "street",
     "lat": 41.3111, "lon": 69.2797, "audio": my_audio, "author": "me"}
)

m = make_map()
m
```

## Step 5 · Save and share

This saves the map as **sound_map.html** and downloads it. Open the file in any browser (Chrome, Firefox, Safari), click a marker and listen. Send it to friends, or put it online with GitHub Pages.

```python
m.save("sound_map.html")
files.download("sound_map.html")
```
